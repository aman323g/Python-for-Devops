Strings 

Strings Manipulation:

strings are ordered, Immutable(not possible to change after creation) sequence of characters.
we can use single, double or triple quotes in strings.
Triple quotes basically used for Multiline or docstrings.

```python
single_line_str = "Double quoted"
single_line_str2= 'single quoted'
print(single_line_str)
print(single_line_str2)
```

Format output using F sting:

It allows inline expression evaluation and formatting, making string construction, concise and readable

```python
math_division = 7/2
print(f"Result: {math_division}")
print(f"Result: {7/2}")
```

Common String operations:
1. Concatenation: Addition of strings
2. length (len()): Gets the number of character
3. indexing ([]): Access a character by position (0-based)
4. slicing ([:]): Extract substrings
5. .lower()/.upper(): used to convert string elements in lower or upper case
6. .strip()/.lstrip()/.rstrip(): used to remove white space from left and right or only left or only right.
7. .startswith()/.endwith()
8. .split()/.join(): split or join the strings
9. .replace()

```python
course_title = "   Python for Devops  "

print(course_title)
print(f"Result of strip: {course_title.strip()}")
print(f"Result of lstrip: {course_title.lstrip()}")
print(f"Result of rstrip: {course_title.rstrip()}")
print(f"Result of upper: {course_title.upper()}")
print(f"Result of lower: {course_title.lower()}")
```

```python
filename = "file.yaml"
print(filename.startswith("file"))
print(filename.endswith("yaml"))
```

Excercise: Calculation of disk usge percentage.

Given the variables below, Compute the disk usage percentage
Print the raw percentage value
Build a human readable summary string
    .convert the server name to upper case
    .include the number of cpu cores and amount of RAM.
    .show the disk usage percentage rounded to one decimal point

Print a summary containing the server name in upper case, the number of cpu cores, the memory and disk usage.
Finally, use the .2% format specifier in an f-string to display the usage with dwo decimal and a percent sign.

Variables:
Server_name = "webserver-03"
cpu_cores = 4
memory_gb = 8.0
disk_total_gb = 500
disk_used_gb = 350

disk_usage_percentage = disk_used_gb/dist_total_gb



```python

Server_name = "webserver-03"
cpu_cores = 4
memory_gb = 8.0
disk_total_gb = 500
disk_used_gb = 350

disk_usage_percentage = disk_used_gb/disk_total_gb

print(disk_usage_percentage) ##Raw value
summary = f"server '{Server_name.upper()}' (CPU: {cpu_cores} {memory_gb}GB RAM) Disk Usage: {disk_usage_percentage}"
print(summary)
summary = f"server '{Server_name.upper()}' (CPU: {cpu_cores} {memory_gb}GB RAM) Disk Usage: {disk_usage_percentage:.2f}" ## round off upto two decimal place
print(summary)
summary = f"server '{Server_name.upper()}' (CPU: {cpu_cores} {memory_gb}GB RAM) Disk Usage: {disk_usage_percentage:.2%}" ## with percentage symbol
print(summary)
```
